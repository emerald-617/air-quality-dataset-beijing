#### Deskripsi Data
Dataset `Air-Quality` merupakan kumpulan data menggambarkan pemantauan kualitas udara di lokasi terkontrol di Tiongkok. Dataset ini mencakup data polutan udara per jam dari 12 lokasi pemantauan kualitas udara yang dikendalikan secara nasional. Data kualitas udara berasal dari Beijing Municipal Environmental Monitoring Center. Data meteorologi di setiap lokasi pemantauan kualitas udara dicocokkan dengan stasiun cuaca terdekat dari China Meteorological Administration. Periode waktunya dari 1 Maret 2013 hingga 28 Februari 2017.

#### Keterangan Kolom
* No: row number
* year: year of data in this row 
* month: month of data in this row 
* day: day of data in this row 
* hour: hour of data in this row 
* PM2.5: PM2.5 concentration (ug/m^3)
* PM10: PM10 concentration (ug/m^3)
* SO2: SO2 concentration (ug/m^3)
* NO2: NO2 concentration (ug/m^3)
* CO: CO concentration (ug/m^3)
* O3: O3 concentration (ug/m^3)
* TEMP: temperature (degree Celsius) 
* PRES: pressure (hPa)
* DEWP: dew point temperature (degree Celsius)
* RAIN: precipitation (mm)
* wd: wind direction
* WSPM: wind speed (m/s)
* station: name of the air-quality monitoring site

### Cleaning Data

Karena semua dataFrame memiliki masalah yang sama, yaitu missing values, maka akan dilakukan penanganan missing values dengan looping semua dataFrame. Penanganan yang dilakukan yaitu imputasi berbasis waktu karena dataset merupakan data time series (runtun waktu). Imputasi berbasis waktu yang digunakan adalah: <br>
- interpolate untuk data numerik (tipe data int dan float)
- forward fill untuk data kategorik (tipe data object)

```python
columns = next(iter(dfs.values())).columns
df_missing = pd.DataFrame({'column_name': columns})

for name, df in dfs.items():
    df_missing[name] = df.isna().sum().values

df_missing.set_index('column_name', inplace=True)
df_missing
```

```python
dfs_clean = {}
for df in dfs.keys():
    num_cols = dfs[df].select_dtypes(exclude="object").interpolate()
    cat_cols = dfs[df].select_dtypes(include="object").ffill()
    dfs_clean[df] = pd.concat([num_cols, cat_cols],axis=1)
```

```python
columns = next(iter(dfs_clean.values())).columns
df_missing = pd.DataFrame({'column_name': columns})

for name, df in dfs_clean.items():
    df_missing[name] = df.isna().sum().values

df_missing.set_index('column_name', inplace=True)
df_missing
```

Seperti yang sudah dideskripsikan pada bagian df_dingling dan df_gucheng, imputasi dengan interpolasi tidak bisa dilakukan karena terdapat missing value dengan posisi di ujung, karena interpolate() hanya mampu menangani missing di tengah. Oleh karena itu, dilakukan imputasi lanjutan dengan ffill/bfill untuk menangani sisa missing value di ujung. Sebelum itu, perlu dicek dulu apakah missing value berada di awal, di akhir, atau di awal dan di akhir untuk kedua dataFrame yang masih ada missing value.

```python
df_cek = dfs_clean["df_gucheng"]
df_cek[df_cek['NO2'].isna()].head(100)
```

```python
df_cek2 = dfs_clean["df_dingling"]
df_cek2[df_cek2['NO2'].isna()].head(100)
```

Missing value berada di awal periode untuk kedua dataframe sehingga imputasi dilakukan dengan backward fill (bfill).

```python
for df in dfs_clean.keys():
    num_cols = dfs_clean[df].select_dtypes(exclude="object").bfill()
    cat_cols = dfs_clean[df].select_dtypes(include="object")
    dfs_clean[df] = pd.concat([num_cols, cat_cols],axis=1)

columns = next(iter(dfs_clean.values())).columns
df_missing = pd.DataFrame({'column_name': columns})

for name, df in dfs_clean.items():
    df_missing[name] = df.isna().sum().values

df_missing.set_index('column_name', inplace=True)
df_missing
```

Karena semua masalah sudah diatasi dan data sudah bersih, maka semua dataFrame dapat di-merge row dengan concat().

```python
df_cleaned = pd.concat(dfs_clean.values(), ignore_index=True)
df_cleaned.head()
```

```python
df_cleaned.info()
```

**Insight:**
- Semua missing value sudah ditangani dengan interpolasi dan backward fill untuk data numerik, forward fill untuk data kategorik.
- Data sudah bersih sehingga bisa disimpan dalam format .csv.

#### Agregasi Harian
Untuk keperluan explanatory analysis, akan dibuat dataFrame baru yang berupa agregasi harian dari df. Ini diperlukan untuk menjawab pertanyaan bisnis yang berkaitan dengan kualitas udara.

Untuk variabel polutan ('PM2.5', 'PM10', 'SO2', 'NO2', 'CO') dan variabel cuaca numerik ('TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM') mengambil nilai rata-rata hariannya. Untuk variabel kategorik ('wd', 'station') mengambil nilai modusnya. Khusus variabel O3, mengambil nilai maksimum rata-rata 8 jam dalam sehari. Hal ini sesuai dengan konsep perhitungan indeks kualitas udara yang akan dijelaskan di bagian berikutnya.

```python
polutan_cols = ['PM2.5', 'PM10', 'SO2', 'NO2', 'CO']
cuaca_cols   = ['TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM']
kategori_cols = ['wd']
agg_dict = {}

def mode_func(x):
    return x.mode().iloc[0] if not x.mode().empty else None

# menggunakan mean untuk variabel polutan & cuaca
for col in polutan_cols + cuaca_cols:
    agg_dict[col] = 'mean'

# menggunakan modus untuk tipe kategorik
for col in kategori_cols:
    agg_dict[col] = mode_func

# agregasi harian
df_agregasi = df.groupby(['station','date']).agg(agg_dict).reset_index()

df = df.sort_values(['station', 'date'])
df['O3_8h_avg'] = df.groupby('station')['O3']\
                    .transform(lambda x: x.rolling(8, min_periods=8).mean())
o3_daily = df.groupby(['station', 'date'])['O3_8h_avg'].max().reset_index()
o3_daily.columns = ['station', 'date', 'O3']

df_agregasi = df_agregasi.merge(o3_daily, on=['station','date'], how='left')
df_agregasi['year']  = df_agregasi['date'].dt.year
df_agregasi['month'] = df_agregasi['date'].dt.month
df_agregasi['day']   = df_agregasi['date'].dt.day
df_agregasi.head()
```

#### AQI (air quality index)
Air Quality Index (AQI) adalah sistem untuk menerjemahkan pengukuran konsentrasi polutan yang terkadang membingungkan atau tidak intuitif, menjadi satu skala yang mudah dipahami dengan jelas mewakili risiko kesehatan yang ditimbulkan oleh polusi udara sekitar. Rumus indeks ini mempertimbangkan 6 polutan utama, yaitu PM2.5, PM10, karbon monoksida, sulfur dioksida, nitrogen dioksida, dan ozon. China dan Amerika memiliki dua sistem yang paling banyak digunakan. Standar China akan digunakan untuk analisis ini karena dataset berlokasi di wilayah-wilayah Tiongkok. AQI China sendiri didasarkan pada protokol pelaporan resmi Cina tetapi menggunakan rata-rata per jam untuk refleksi yang lebih segera dari tingkat polusi udara.

**China AQI Level**: <br>
Excellent            -> 0–50    <br>
Good                 -> 51–100  <br>
Lightly Polluted     -> 101–150 <br>
Moderately Polluted  -> 151–200 <br>
Heavily Polluted     -> 201–300 <br>
Severely Polluted    -> 301+    <br>

**Rumus AQI**:
$$
I_P = \frac{I_{high}-I_{low}}{BP_{high}-BP_{low}} * (C_P-BP_{low}) + I_{low}
$$

$$AQI=Max(I_{PM2.5}, I_{PM10}, I_{SO2}, I_{NO2}, I_{CO}, I_{O3})$$

Breakpoints masing-masing variabel polutan didasarkan pada https://aqihub.info/indices/china yang bersumber dari Technical Regulation on Ambient Air Quality Index (on trial).

**Sumber**: <br>
- https://www.iqair.com/id/support/knowledge-base/what-is-the-difference-between-china-us-aqi
- https://aqihub.info/indices/china

```python
# mendefinisikan breakpoints
bp_pm25 = [
    (0, 35, 0, 50),
    (35, 75, 50, 100),
    (75, 115, 100, 150),
    (115, 150, 150, 200),
    (150, 250, 200, 300),
    (250, 500, 300, 600),
]

bp_pm10 = [
    (0, 50, 0, 50),
    (50, 150, 50, 100),
    (150, 250, 100, 150),
    (250, 350, 150, 200),
    (350, 420, 200, 300),
    (420, 500, 300, 600)
]

bp_so2 = [
    (0, 50, 0, 50),
    (50, 150, 50, 100),
    (150, 475, 100, 150),
    (475, 800, 150, 200),
    (800, 1600, 200, 300),
    (1600, 2100, 300, 500)
]

bp_no2 = [
    (0, 40, 0, 50),
    (40, 80, 50, 100),
    (80, 180, 100, 150),
    (180, 280, 150, 200),
    (280, 565, 200, 300),
    (565, 750, 300, 500)
]

bp_co = [
    (0, 2000, 0, 50),
    (2000, 4000, 50, 100),
    (4000, 14000, 100, 150),
    (14000, 24000, 150, 200),
    (24000, 36000, 200, 300),
    (36000, 40000, 300, 500)
]

bp_o3 = [
    (0, 100, 0, 50),
    (101, 160, 50, 100),
    (160, 215, 100, 150),
    (215, 265, 150, 200),
    (265, 800, 200, 300),
    (800, 1100, 300, 500)
]

# rumus dasar iaqi
def calc_iaqi(C, breakpoints):
    for C_low, C_high, I_low, I_high in breakpoints:
        if C_low <= C <= C_high:
            return ((I_high - I_low) / (C_high - C_low)) * (C - C_low) + I_low
    return None
```

Setelah mendefinisikan breakpoints masing-masing variabel dan membuat fungsi untuk rumus dasar aqi, langkah selanjutnya menghitung nilai IAQI masing-masing variabel polutan. Nilai AQI ditentukan berdasarkan nilai IAQI terbesar di antara semua variabel polutan (IAQI max) pada baris yang sama kemudian variabel yang memiliki nilai max tersebut dimasukan ke kolom baru yang disebut AQI_parameter.

```python
df_aqi = df_agregasi.copy()
df_aqi['IAQI_PM2.5'] = df_aqi['PM2.5'].apply(lambda x: calc_iaqi(x, bp_pm25))
df_aqi['IAQI_PM10'] = df_aqi['PM10'].apply(lambda x: calc_iaqi(x, bp_pm10))
df_aqi['IAQI_SO2'] = df_aqi['SO2'].apply(lambda x: calc_iaqi(x, bp_so2))
df_aqi['IAQI_NO2'] = df_aqi['NO2'].apply(lambda x: calc_iaqi(x, bp_no2))
df_aqi['IAQI_CO'] = df_aqi['CO'].apply(lambda x: calc_iaqi(x, bp_co))
df_aqi['IAQI_O3'] = df_aqi['O3'].apply(lambda x: calc_iaqi(x, bp_o3))

cols = [
    'IAQI_PM2.5',
    'IAQI_PM10',
    'IAQI_SO2',
    'IAQI_NO2',
    'IAQI_CO',
    'IAQI_O3'
]

df_agregasi['AQI'] = df_aqi[cols].max(axis=1)

df_agregasi['AQI_parameter'] = (
    df_aqi[cols]
    .idxmax(axis=1)
    .str.replace('IAQI_', '', regex=False)
)

df_agregasi.head()
```

```python
df_aqi[cols].head()
```

Contoh perhitungan AQI → misalnya pada row 0 nilai terbesar ada pada IAQI_O3, yaitu 37.25. Maka nilai AQI pada row 0 adalah 37.25 dan AQI_parameter pada row 0 adalah O3. Dan seterusnya.

Sedikit tambahan informasi:
- Breakpoints AQI hourly (1 hr.) dan daily (24 hr.) berbeda sehingga nilai AQI hourly tidak selalu bisa digunakan untuk menghitung AQI daily dengan cara agregasi harian. 
- Nilai breakpoints yang sebelumnya didefinisikan didasarkan pada breakpoints AQI daily (24 hr., khusus O3 8 hr.) sesuai pada sumber https://aqihub.info/indices/china. 
- Nilai breakpoint CO sedikit berbeda antara di web dan file ini karena perbedaan satuan (satuan di web dikonversi menyesuaikan dataset yang diberikan).

Untuk menjawab pertanyaan tersebut, diperlukan dataFrame yang mencakup rata-rata bulanan semua stasiun untuk `AQI` dan variabel-variabel polutan, yaitu `PM2.5`, `PM10`, `SO2`, `NO2`, `CO`, `O3`.

```python
polutan_aqi_cols = ['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3', 'AQI']
df_agregasi['year_month'] = df_agregasi['date'].dt.to_period('M')

df_monthly_station = (
    df_agregasi.groupby(['station', 'year_month'])[polutan_aqi_cols]
    .mean()
    .reset_index()
)
df_monthly_station['year_month'] = df_monthly_station['year_month'].dt.to_timestamp()
df_monthly_station.rename(columns={'year_month': 'date'}, inplace=True)
df_monthly_station = df_monthly_station.sort_values(by=["station","date"]).reset_index()
df_monthly_station.drop(columns="index", inplace=True)

df_monthly_station.head()
```

```python
aqi_levels = [
    (50, "Excellent"),
    (100, "Good"),
    (150, "Lightly Polluted"),
    (200, "Moderately Polluted"),
    (300, "Heavily Polluted"),
]
colors = ['green', 'lime', 'orange', 'red', 'purple']

stations = df_monthly_station.station.unique()

fig, axes = plt.subplots(6, 2, figsize=(15, 12), sharex=True)
axes = axes.flatten()

for i, st in enumerate(stations):
    df_st = df_monthly_station[df_monthly_station['station'] == st]
    
    axes[i].plot(df_st['date'], df_st['AQI'], marker='o')
    axes[i].set_title(f"Time Series Rata-rata Bulanan AQI Stasiun {st}")
    axes[i].set_ylabel("AQI")

    # garis horizontal
    for j, (level, label) in enumerate(aqi_levels):
        axes[i].axhline(y=level, linestyle='--', color=colors[j], alpha=0.7)

legend_lines = [
    Line2D([0], [0], color=colors[i], lw=1.5, linestyle='--', label=label)
    for i, (_, label) in enumerate(aqi_levels)
]
fig.legend(
    handles=legend_lines,
    title="AQI Category",
    loc='lower center',
    bbox_to_anchor=(0.5, -0.02),
    ncol=5
)

plt.xlabel("Waktu")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Analisis Lanjutan (Opsional): Binning

```python
bins = [0, 50, 100, 150, 200, 300, float('inf')]

labels = [
    'Excellent',
    'Good',
    'Lightly Polluted',
    'Moderately Polluted',
    'Heavily Polluted',
    'Severely Polluted'
]

df_monthly_station['AQI_cat'] = pd.cut(
    df_monthly_station['AQI'],
    bins=bins,
    labels=labels,
    right=True
)
df_monthly_station.head()
```

```python
fig, axes = plt.subplots(4, 3, figsize=(15, 12))
axes = axes.flatten()

colors_order = ["#D3D3D3", "#D3D3D3", "#5774B7", "#D3D3D3", "#D3D3D3", "#D3D3D3"]

for i, st in enumerate(stations):
    df_st = df_monthly_station[df_monthly_station['station'] == st]
    
    dist = df_st['AQI_cat'].value_counts(normalize=True) * 100
    dist = dist.reindex(labels, fill_value=0)
    
    axes[i].barh(dist.index, dist.values, color=colors_order)
    axes[i].set_title(st)
    axes[i].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

**Insights**:
- Nilai AQI di 12 stasiun berfluktuasi siklikal yang memperlihatkan pola musiman.
- Pada plot time series, terlihat bahwa hampir semua stasiun berada di bawah garis kuning atau dengan kata lain hampir semua stasiun berada pada kategori `Lightly Polluted` secara keseluruhan di sepanjang 2013-2017. Hal ini terlihat lebih jelas pada barchart dengan semua stasiun memiliki kategori `Lightly Polluted` sebagai hari terbanyak di sepanjang 2013-2017

- Stasiun Changping, Dingling, dan Huairou beberapa kali hampir memasuki kateogri `Excellent`. Bahkan pada barchart terlihat bahwa ketiga stasiun ini memiliki hari dengan kategori `Good` yang cukup banyak. Kemungkinan area-area tersebut cenderung lebih bersih dibanding area lain.
- Stasiun Dongsi, Gucheng, Nongzhanguan, Shunyi, Tiantan, Wanliu, dan Wanshouxigong lebih sering melebihi garis merah atau masuk ke kategori `Heavily Polluted`. Area-area tersebut dapat menjadi prioritas kebijakan.

```python
df_monthly_2013 = df_monthly_station[df_monthly_station['date'].dt.year == 2013]
df_monthly_2017 = df_monthly_station[df_monthly_station['date'].dt.year == 2017]

aqi_2013 = df_monthly_2013.groupby('station')['AQI'].mean()
aqi_2017 = df_monthly_2017.groupby('station')['AQI'].mean()

df_delta = pd.DataFrame({
    'AQI_2013': aqi_2013,
    'AQI_2017': aqi_2017
})

df_delta['delta_AQI'] = df_delta['AQI_2017'] - df_delta['AQI_2013']

df_delta['pct_change'] = (
    (df_delta['AQI_2017'] - df_delta['AQI_2013']) /
    df_delta['AQI_2013']
) * 100

df_delta = df_delta.sort_values(by='pct_change', ascending=True)

plt.figure(figsize=(10, 6))

bars = plt.barh(
    df_delta.index,
    df_delta['pct_change'],
    color=df_delta['pct_change'].apply(lambda x: 'green' if x < 0 else 'red')
)

plt.axvline(0, color='black', linewidth=1)

plt.xlabel('Perubahan AQI (%) (2017 vs 2013)')
plt.title('Perbandingan Relatif AQI per Stasiun')

for i, v in enumerate(df_delta['pct_change']):
    plt.text(
        v,
        i,
        f'{v:.1f}%',
        va='center',
        ha='left' if v > 0 else 'right'
    )

plt.tight_layout()
plt.show()
```

**Insights**:
- Dari 12 stasiun, **8 stasiun memburuk dan hanya 4 stasiun membaik** dalam periode 2013-2017. Hal ini berarti, secara keseluruhan, kebijakan kualitas udara di Beijing belum efektif merata.
- Stasiun **Shunyi (+11.9%), Gucheng (+11.3%), dan Wanshouxigong (+11.1%)** merupakan 3 terburuk dengan peningkatan AQI dari 2013 ke 2017 hampir 12%. Angka ini cukup signifikan dalam 4 tahun.
- Stasiun  **Changping (-5.4%), Dingling (-4.9%), dan Huairou (-3.5%)** menunjukkan perbaikan terbesar. Wanliu juga membaik, namun persentase hanya -0.9%, hampir tidak signfikan.

#### Proporsi Bulanan

```python
df_monthly_all['AQI_cat'] = pd.cut(
    df_monthly_all['AQI'],
    bins=bins,
    labels=labels,
    right=True
)

df_monthly_all['year'] = df_monthly_all['date'].dt.year
df_monthly_all['month'] = df_monthly_all['date'].dt.month

monthly_dist = pd.crosstab(
    df_monthly_all['month'],
    df_monthly_all['AQI_cat'],
    normalize='index'
) * 100

monthly_dist = monthly_dist.reindex(columns=labels, fill_value=0)

ax = monthly_dist.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6),
    color=['green', 'lime', 'orange', 'red', 'purple', 'black']
)

for i, row in enumerate(monthly_dist.values):
    cumulative = 0
    for j, val in enumerate(row):
        if val > 0:
            ax.text(
                i, 
                cumulative + val / 2,
                f"{val:.1f}%",
                ha='center',
                va='center',
                fontsize=8
            )
        cumulative += val

plt.title("Distribusi Kategori AQI per Bulan (%)")
plt.xlabel("Bulan")
plt.ylabel("Persentase (%)")
plt.legend(title="AQI Category", bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.xticks(rotation=0)
plt.show()
```

#### Proporsi Tahunan

```python
yearly_dist = pd.crosstab(
    df_monthly_all['year'],
    df_monthly_all['AQI_cat'],
    normalize='index'
) * 100

yearly_dist = yearly_dist.reindex(columns=labels, fill_value=0)

ax = yearly_dist.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6),
    color=['green', 'lime', 'orange', 'red', 'purple']
)

for i, row in enumerate(yearly_dist.values):
    cumulative = 0
    for j, val in enumerate(row):
        if val > 0:
            ax.text(
                i, 
                cumulative + val / 2,
                f"{val:.1f}%",
                ha='center',
                va='center',
                fontsize=8
            )
        cumulative += val

plt.title("Distribusi Kategori AQI per Tahun (%)")
plt.xlabel("Tahun")
plt.ylabel("Persentase (%)")
plt.legend(title="AQI Category", bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.xticks(rotation=0)
plt.figtext(
    0.5, -0.05,
    "*2013: Mar–Des | *2017: Jan–Feb",
    ha='center',
    fontsize=9
)

plt.show()
```

**Insights:**
- Nilai AQI fluktuatif siklikal sepanjang periode pengamatan yang menunjukkan adanya **pola musiman** untuk keseluruhan stasiun Beijing.
- Plot time series dan stacked barchart menunjukkan bahwa kategori `Lightly Poluted` mendominasi hampir di semua bulan pengamatan, menunjukkan bahwa kualitas udara di Beijing secara bulanan selama 2013-2017 cukup berpolusi. Hal tersebut juga berlaku untuk level tahunan yang berarti **Beijing secara konsisten berada di kategori tidak sehat.**
- Kategori `Moderately Polluted` terlihat cukup banyak pada bulan Januari, Februari, dan Desember atau pada periode Musim Dingin. Hal ini berarti **kualitas udara memburuk pada Musim Dingin** sehingga kebijakan dapat difokuskan pada musim tersebut.
- Proporsi pada **2016 sedikit berbeda dari tahun-tahun lainnya**, di mana kategori AQI `Good` jauh lebih tinggi dari tahun lain, yaitu **33.3%**.
- Kondisi kualitas udara pada 2017 terlihat cukup berbahaya dengan proporsi terbanyak ada pada kategori `Moderately Polluted`. Namun, hasil ini mungkin **misleading** karena dataset hanya mencakup bulan Januari dan Februari pada tahun 2017. Selain itu, pada plot time series bulanan diketahui bahwa kualitas udara memburuk pada periode Musim Dingin sehingga sangat mungkin proporsi kualitas udara pada 2017 terlihat buruk karena hanya mencakup bulan-bulan pada Musim Dingin.

### Pertanyaan 2: Bagaimana perubahan pola musiman variabel polutan (PM2.5, PM10, SO2, NO2, CO, O3) antar musim selama 2013-2017 dan apakah perubahan pola konsisten di seluruh 12 stasiun?

Untuk menjawab pertanyaan tersebut, diperlukan dataFrame agregasi mean berdasarkan season untuk setiap stasiun. Fitur season dibuat dengan menggunakan mapping.

Fitur season mengacu pada musim di Tiongkok, yaitu:
1. Spring -> Maret, April, Mei
2. Summer -> Juni, Juli, Agustus
3. Autumn -> September, Oktober, November
4. Winter -> Desember, Januari, Februari

```python
daily_cols = ['station', 'date', 'year', 'month', 'day', 'PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3', 'TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM', 'AQI', 'AQI_parameter']
df_daily_station = df_agregasi[daily_cols].copy()

season_mapping = {
    3: 'Spring', 4: 'Spring', 5: 'Spring',
    6: 'Summer', 7: 'Summer', 8: 'Summer',
    9: 'Autumn', 10: 'Autumn', 11: 'Autumn',
    12: 'Winter', 1: 'Winter', 2: 'Winter'
}

df_daily_station["season"] = df_daily_station["month"].map(season_mapping)
df_daily_station.sample(7)
```

```python
polutan_cols.append('O3')

df_season_station = (
    df_daily_station.groupby(['station','season'])[polutan_cols]
    .mean()
    .reset_index()
)

df_season_station
```

```python
station_order = df_season_station.station.unique()
season_order = ['Spring', 'Summer', 'Autumn', 'Winter']

fig, axes = plt.subplots(3, 2, figsize=(14, 20))
axes = axes.flatten()

for i, pol in enumerate(polutan_cols):
    pivot = df_season_station.pivot(
        index='station',
        columns='season',
        values=pol
    ).reindex(index=station_order, columns=season_order)
    
    sns.heatmap(
        pivot,
        ax=axes[i],
        cmap='RdYlGn_r',
        annot=True,
        fmt='.1f',
        cbar=True
    )
    
    axes[i].set_title(f'{pol}')
    axes[i].set_xlabel('')
    axes[i].set_ylabel('')

plt.tight_layout()
plt.show()
```

**Insights**:
- `PM2.5` menunjukkan pola yang konsisten di mana konsentrasi **tertinggi di Winter (merah), terendah di Summer (hijau)**.
- `PM10` menunjukkan konsentrasi **tertinggi di Spring, terendah di Summer**. Hal ini berarti terjadi penurunan konsentrasi yang signfikan dari Spring ke Summer.
- **`SO2` mempunyai pola serupa dengan `PM2.5` dengan gradasi lebih ekstrem**. Konsentrasi tertinggi di Winter dan terendah di Summer.
- Konsentrasi `NO2` pada **Winter cukup tinggi, tetapi pada Summer konsentrasinya tidak serendah `SO2`**.
- `CO2` dibandingkan polutan lain, **nilai absolut jauh lebih besar yaitu 800-2100 ug/m^3**. Pola pada Winter konsisten tertinggi, dengan nilai tertinggi ada pada stasiun Gucheng Winter, yaitu 2106.0. Perlu analisis lebih lanjut apa penyebabnya.
- **Pola `O3` terbalik dibanding semua polutan lain: Summer tertinggi dan Winter terendah**. Stasiun Nongzhaguan mencatat konsentrasi O3 Summer tertinggi, yaitu 172.5.
- Secara umum, **semua stasiun memiliki pola sangat konsisten untuk semua polutan** yang ditunjukkan oleh warna di setiap kolom musim relatif seragam.

### Pertanyaan 3: Bagaimana kuat korelasi variabel cuaca (TEMP, PRES, DEWP, WSPM, RAIN) terhadap nilai rata-rata AQI harian di 12 stasiun pemantauan selama periode 2013-2017?

Untuk menjawab pertanyaan tersebut, diperlukan dataFrame yang mencakup rata-rata semua stasiun untuk setiap variabel cuaca dan variabel AQI.

```python
daily_all_cols = ['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3', 'TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM', 'AQI']
df_daily_all = (
    df_daily_station.groupby('date')[daily_all_cols]
    .mean()
    .reset_index()
)
df_daily_all.head()
```

```python
cuaca_cols
```

```python
corr_matrix = df_daily_all[['AQI'] + cuaca_cols].corr()
correlation = corr_matrix.loc['AQI', cuaca_cols]
correlation = correlation.to_frame().T

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap='coolwarm',
    center=0,
    linewidth=0.5,
)

plt.title("Heatmap Korelasi Cuaca vs. AQI")
plt.xlabel("Variabel Cuaca")
plt.tight_layout()
plt.show()
```

```python
n_cols = 2
n_rows = (len(cuaca_cols) + 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(12, 8))
axes = axes.flatten()

for i, col in enumerate(cuaca_cols):
    sns.regplot(
        x=col,
        y='AQI',
        data=df_daily_all,
        ax=axes[i],
        scatter_kws={'alpha': 0.4},
        line_kws={'color': 'red'}
    )
    
    axes[i].set_title(f'AQI vs {col}')

for j in range(len(cuaca_cols), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

**Insights**:
- `TEMP`, `PRES`, `DEWP`, `RAIN` hampir tidak berkorelasi atau berkolerasi lemah terhadap `AQI` dengan nilai antara **-0.09 hingga +0.18**. Sementara itu, `WSMP` memiliki korelasi sedang negatif, yaitu **-0.42**.
- Scatterplot juga menunjukkan hal yang sama. Garis tren pada `TEMP`, `PRES`, dan `DEWP` **hampir datar dan tidak menunjukkan adanya hubungan linier sama sekali**. 
- Scatterplot `AQI` vs `RAIN` memiliki distribusi data yang **sangat skewed**, hal ini terjadi karena mayoritas nilai RAIN adalah 0 sehingga data bertumpuk di kiri.
- Tren negatif paling terlihat jelas secara visual pada scatterplot `AQI` vs `WSPM`. **Semakin tinggi kecepatan angin (wspm), semakin rendah `AQI`**.

**Catatan Keterbatasan Analisis**: <br>
Visualisasi heatmap dan scatterplot hanya menunjukkan korelasi linier sehingga tidak menutup kemungkinan variabel-variabel dengan korelasi rendah seperti `TEMP` sebenarnya memiliki pola hubungan non-linier atau pola hubungan yang tidak dapat ditangkap hanya dengan heatmap atau scatterplot. Sehingga perlu dilakukan analisis lebih lanjut menggunakan model yang lebih kompleks. Sedikit tambahan, heatmap menggunakan korelasi Pearson menghasilkan output yang lebih baik pada data berdistribusi normal. Sementara itu, pada scatterplot dapat terlihat beberapa variabel menunjukkan distribusi skewed yang sangat kentara sehingga hasil heatmap kurang representatif.
